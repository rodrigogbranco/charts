```python
import plotly.express as px
import plotly.io as pio
```

```python
# Sample data
df = pd.DataFrame({
"Cenário": ["Origem=Destino", "Origem=Destino", 
            "Origem=Destino", "Sem obstáculos", 
            "Sem obstáculos", "Sem obstáculos", 
            "Situação típica", "Situação típica", "Situação típica"],
"Métricas": [0, 1, 2, 0, 1, 2, 0, 1, 2],
"Tempo (s)": [0, 0, 0, 0, 70, 70, 30, 70, 100]
})

# Aggregate by Pclass and Survived
grouped_df = df.groupby(["Cenário", "Métricas"], as_index=False).agg({"Tempo (s)": "sum"})

# Convert numeric category to string for discrete coloring
grouped_df["Métricas"] = grouped_df["Métricas"].map({0: "tl", 1: "btt", 2: "ttt"})
```

```python
fig = px.bar(
grouped_df,
x="Cenário",
y="Tempo (s)",
color="Métricas",
barmode="group",
title="Comportamento das métricas tl, btt e ttt"
)

fig.update_layout(yaxis={'showticklabels': False})

fig.show()
```

```python
pio.write_image(
        fig,
        "{}/{}.pdf".format("figs-tese-2025-06-28", "btt-ttt-tl"),
        format="pdf",
        width=800,
        height=480,
    )
```
